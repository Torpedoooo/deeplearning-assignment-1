```python
import os
import pandas as pd
from PIL import Image
from torch.utils.data import Dataset
import matplotlib.pyplot as plt
from torchvision import transforms
from torch.utils.data import DataLoader
from torch.utils.data import WeightedRandomSampler
from sklearn.metrics import classification_report, confusion_matrix, f1_score
import torch
import torch.nn as nn
import seaborn as sns
import random
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')


class CustomImageDataset(Dataset):
    def __init__(self, annotations_file, img_dir,
                 transform=None, target_transform=None):
        self.img_labels = pd.read_csv(annotations_file)
        classes = sorted(self.img_labels.iloc[:,1].unique())
        self.class2idx = {c:i for i,c in enumerate(classes)}
        self.img_dir    = img_dir
        self.transform  = transform
        self.target_transform = target_transform

    def __len__(self):
        return len(self.img_labels)

    def __getitem__(self, idx):
        if torch.is_tensor(idx):
            idx = idx.tolist()

        img_name = self.img_labels.iloc[idx, 0]            # first column = filename
        if not img_name.lower().endswith('.png'):
            img_name = img_name + '.png'
        label = self.img_labels.iloc[idx, 1]            # second column = class
        label = self.class2idx[label]
        img_path = os.path.join(self.img_dir, img_name)
        image = Image.open(img_path).convert("RGB")

        if self.transform:
            image = self.transform(image)
        if self.target_transform:
            label = self.target_transform(label)

        return image, label
```

### Data Preprocessing & Augmentation Strategy

To prepare the Pokémon images for our Multilayer Perceptron (MLP), we implemented a specific transformation pipeline:
* **Resizing (64x64):** MLPs require flattened 1D arrays, meaning the number of input features scales quadratically with image dimensions. Resizing to 64x64 keeps the parameter count manageable while retaining enough visual detail for basic classification.
* **Data Augmentation (Random Horizontal Flip):** We applied a simple 50% chance to flip the image horizontally. This artificially increases the diversity of our training data and helps prevent overfitting. **Note:** We deliberately avoided heavy spatial augmentations (like rotation or translation). Because MLPs flatten images into 1D vectors, they rely on pixels being in exact, fixed locations. Heavy rotations destroy this rigid structure and confuse the MLP.
* **Normalization:** We normalized the pixel values using standard ImageNet statistics (Mean: `[0.485, 0.456, 0.406]`, Std: `[0.229, 0.224, 0.225]`) to center the data, ensuring stable and faster gradient descent.

```python
# build a transformation that will be applied to every image
train_transform = transforms.Compose([
    transforms.Resize((64, 64)),    # make all samples the same size
    transforms.RandomHorizontalFlip(),# data‑augmentation
    transforms.ToTensor(),
    transforms.Normalize(             # zero‑mean, unit‑variance per channel
        mean=[0.485, 0.456, 0.406],
        std =[0.229, 0.224, 0.225])
])

val_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406],
                         std =[0.229,0.224,0.225])
])
```

```python
df = pd.read_csv('train_labels.csv')

# shuffle & take 80 % for training
train_df = df.sample(frac=0.8, random_state=42)

# the remainder 20%0  is the validation set
test_df  = df.drop(train_df.index)

train_df.to_csv('train_split.csv', index=False)
test_df.to_csv('test_split.csv',  index=False)

print("train:", len(train_df), "rows")
print("test :", len(test_df),  "rows")
```

```python

train_csv = './train_labels.csv'
train_dir = './Train/'

train_ds = CustomImageDataset('train_split.csv', train_dir, transform=train_transform)
test_ds  = CustomImageDataset('test_split.csv', train_dir, transform=val_transform)


# numeric labels for every train example
labels = train_ds.img_labels['label'].map(train_ds.class2idx)

#class‑level weights
class_counts  = labels.value_counts().sort_index()
class_weights = 1.0 / class_counts
class_weights = class_weights / class_weights.sum()

# map the weight array
sample_weights = labels.map(class_weights).values
sample_weights = torch.tensor(sample_weights, dtype=torch.double)

#sampler that will draw from the dataset with those probabilities
sampler = WeightedRandomSampler(
    weights = sample_weights,
    num_samples = len(sample_weights),   # one draw per original example
    replacement = True                   # allow repeats to keep using less frequent labelks
)

train_loader = DataLoader(
    train_ds,
    batch_size=64,
    sampler=sampler,

)
print("split counts:\n", train_ds.img_labels['label'].value_counts())

for i in range(5):
    xb, yb = next(iter(train_loader))
    print("batch", i, torch.bincount(yb))
test_loader = DataLoader(test_ds, batch_size=32, shuffle=False)
```

### Label Distribution & Class Imbalance Handling

As shown in the plot above, our dataset suffers from severe class imbalance. The **Water** type has nearly three times as many examples as the rarest types (**Ground**, **Rock**, **Fighting**). 

If we train a naive classifier with standard cross-entropy loss, the model will heavily favor the majority classes to artificially inflate its overall accuracy, while completely failing to recognize the minority classes. 

**Our Solution: Oversampling via `WeightedRandomSampler`**
Instead of altering the loss function, we manipulated the data loader. We calculated a weight for each class inversely proportional to its frequency. By passing these weights to a PyTorch `WeightedRandomSampler`, we ensure that during training, the DataLoader pulls minority classes more frequently. This guarantees that every batch the model sees has a roughly equal representation of all 9 Pokémon types, preventing the model from becoming biased towards Water or Normal types.

### Model Architecture: Multilayer Perceptron (MLP)

For Stage 1 of this assignment, we are restricted to using a Multilayer Perceptron. 

**Architecture Choices:**
* **Input Layer:** We use `nn.Flatten()` to convert the 3D image tensors (Channels x Height x Width) into 1D feature vectors.
* **Hidden Layers (`[1024, 512, 256]`):** We designed a large, deep network. Because MLPs cannot detect spatial 2D patterns (like shapes or edges), they must brute-force memorize raw pixel distributions. Providing a large number of parameters gives the network the capacity to learn these complex mapping functions.
* **Regularization (`Dropout`):** Because our network is so large, it is highly prone to overfitting (memorizing the training data perfectly while failing on validation). We added `nn.Dropout(0.3)` after our activation functions to randomly turn off 30% of the neurons during training. This forces the network to learn robust, redundant features rather than relying on a few specific pixel pathways.

```python
criterion = nn.CrossEntropyLoss()
```

```python
class MLP(nn.Module):
    def __init__(self, num_outputs, num_hiddens, dropout_rate=0.3):
        super().__init__()

        # num_hiddens can be a single int or a list of ints
        if isinstance(num_hiddens, int):
            num_hiddens = [num_hiddens]

        layers = [nn.Flatten()]                    # flatten from C×H×W to vector

        for h in num_hiddens:
            layers.append(nn.LazyLinear(h))        # size inferred on first forward
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(dropout_rate))

        layers.append(nn.LazyLinear(num_outputs))  # final classification layer
        self.net = nn.Sequential(*layers)

    def forward(self, X):
        return self.net(X)
```

### Training Strategy & Resource Optimization

To train the model efficiently and satisfy the computational budget requirements, we implemented the following strategies:
* **Optimizer & Scheduler:** We use the Adam optimizer for fast, adaptive learning, paired with a `StepLR` scheduler that reduces the learning rate by half every 20 epochs to help the model settle into the global minimum smoothly.
* **Custom Early Stopping (Macro F1-Score):** Instead of saving the model with the lowest validation loss or highest raw accuracy, we configured our Early Stopping mechanism to monitor the **Validation Macro F1-Score**. Because our dataset is inherently imbalanced, raw accuracy can be a deceptive metric. The Macro F1-Score calculates the performance for each class individually and averages them, ensuring we only save the model if it is genuinely learning to classify *all* Pokémon types, not just the common ones. Training will automatically halt if the F1-Score does not improve for 25 epochs, saving critical GPU compute time.

```python
model = MLP(num_outputs=9, num_hiddens=[1024, 512, 256]).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=20, gamma=0.5)
# helper to compute accuracy on a batch
def batch_accuracy(logits, labels):
    preds = logits.argmax(dim=1)
    return (preds == labels).float().mean().item()
best_f1 = 0.0
trigger_times = 0
patience = 25
epochs = 150
for epoch in range(1, epochs+1):
    model.train()
    running_loss = 0.0
    running_acc  = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)

        logits = model(xb)               # forward pass
        loss = criterion(logits, yb)     # weighted cross‑entropy

        optimizer.zero_grad()
        loss.backward()                  # compute gradients
        optimizer.step()                 # update weights

        running_loss += loss.item() * xb.size(0)
        running_acc  += batch_accuracy(logits, yb) * xb.size(0)
    scheduler.step()
    train_loss = running_loss / len(train_ds)
    train_acc  = running_acc  / len(train_ds)

    # validation
    model.eval()
    val_loss = 0.0
    val_acc  = 0.0
    all_preds = []
    all_targets = []
    with torch.no_grad():
        for xb, yb in test_loader:
            xb, yb = xb.to(device), yb.to(device)
            logits = model(xb)
            val_loss += criterion(logits, yb).item() * xb.size(0)
            val_acc  += batch_accuracy(logits, yb) * xb.size(0)
            preds = logits.argmax(dim=1)
            all_preds.append(preds.cpu())
            all_targets.append(yb.cpu())
    final_preds   = torch.cat(all_preds)
    final_targets = torch.cat(all_targets)
    current_macro_f1 = f1_score(final_preds, final_targets, average='macro')
    if current_macro_f1 > best_f1:
        best_f1 = current_macro_f1
        trigger_times = 0
        torch.save(model.state_dict(), 'best_model.pth')
        print(f"Epoch {epoch:2d}: New best f1_score! Saving model...")
    else:
        trigger_times += 1
        print(f"Epoch {epoch:2d}: No improvement. (Patience: {trigger_times}/{patience})")

        if trigger_times >= patience:
            print("Early stopping triggered! Cutting training short.")
            break
    val_loss /= len(test_ds)
    val_acc  /= len(test_ds)

    print(f"Epoch {epoch:2d} | "
          f"train loss {train_loss:.4f} acc {train_acc:.3f} | "
          f"val loss {val_loss:.4f} acc {val_acc:.3f} | "
          f"macro f1_score {current_macro_f1: .3f}")
```

```python
# --- evaluation on validation set ------------------------------------------
model.load_state_dict(torch.load('best_model.pth'))  # load the best model
model.eval()
all_preds, all_targets = [], []
with torch.no_grad():
    for xb, yb in test_loader:
        xb, yb = xb.to(device), yb.to(device)
        logits = model(xb)
        preds = logits.argmax(dim=1)
        all_preds.append(preds.cpu())
        all_targets.append(yb.cpu())

all_preds   = torch.cat(all_preds)
all_targets = torch.cat(all_targets)

# overall accuracy
accuracy = (all_preds == all_targets).float().mean().item()
print(f"validation accuracy: {accuracy:.4f}")

# per‑class metrics
type_names = sorted(train_ds.class2idx, key=lambda k: train_ds.class2idx[k])
print(classification_report(all_targets, all_preds,
                            target_names=type_names))

# confusion matrix
cm = confusion_matrix(all_targets, all_preds)
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt="d", xticklabels=type_names,
            yticklabels=type_names, cmap="Blues")
plt.xlabel("predicted"); plt.ylabel("true"); plt.title("Confusion matrix")
plt.show()
```

### Task 1 Conclusions & Limitations

By observing the classification report and confusion matrix, we can see the fundamental limitations of using an MLP for image classification:

1. **The Spatial Awareness Problem:** The accuracy plateau is expected. By flattening a 64x64 image, we completely destroy the 2D spatial relationships between pixels. The model cannot understand what a "tail", "shell", or "leaf" looks like; it only knows the exact positions of specific color values. 
2. **Overfitting Wall:** Despite our `Dropout` regularization and `WeightedRandomSampler`, MLPs struggle to generalize visual data. Once the network memorizes the primary training images, validation performance flatlines.
3. **Imbalance Impact:** Thanks to our Macro F1 tracking and weighted sampling, the model makes an active attempt to predict minority classes, rather than just guessing "Water" for every image. 

**Next Steps (Task 2):** To achieve a much higher accuracy, we must transition to a Convolutional Neural Network (CNN). A CNN will slide 2D filters across the images, allowing it to preserve spatial geometry, recognize actual physical features of the Pokémon, and drastically reduce the total number of parameters required.

```python
idx2class = {v:k for k,v in train_ds.class2idx.items()}
test_ids = [f.split('.png')[0] for f in os.listdir('Test') if f.endswith('.png')]

class ImageOnlyDataset(Dataset):
    def __init__(self, ids, img_dir, transform=None):
        self.ids = ids
        self.img_dir = img_dir
        self.transform = transform

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, idx):
        name = self.ids[idx]
        path = os.path.join(self.img_dir, name + '.png')
        img = Image.open(path).convert('RGB')
        if self.transform:
            img = self.transform(img)
        return img

test_ds = ImageOnlyDataset(test_ids, 'Test/', transform=val_transform)
val_loader = DataLoader(test_ds, batch_size=64, shuffle=False)

# prediction loop adjusts
preds = []
with torch.no_grad():
    for xb in val_loader:
        xb = xb.to(device)
        logits = model(xb)
        preds.extend(logits.argmax(1).cpu().tolist())
print("test prediction distribution:", torch.bincount(torch.tensor(preds)))
assert len(preds) == len(test_ids)

# map back to strings and write submission file
out_df = pd.DataFrame({
    'Id': test_ids,
    'label': [idx2class[i] for i in preds]
})
out_df.to_csv('submission2.csv', index=False)
print('wrote', len(out_df), 'predictions to submission.csv')
```
